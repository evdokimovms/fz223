```python
from pathlib import Path
import json

try:
    import chardet
except ImportError:
    print("Установите: pip install chardet")
    raise


RAW_DIR = Path("./miptdoc")                 
OUT_DIR = Path("./mipt_tz_dataset/internal_processed")

OUT_DIR.mkdir(parents=True, exist_ok=True)


def read_any_encoding(path: Path) -> str:
    data = path.read_bytes()
    enc = chardet.detect(data)["encoding"]
    if enc is None:
        enc = "utf-8"
    try:
        return data.decode(enc)
    except Exception:
        for e in ["utf-8", "cp1251", "windows-1251", "utf-8-sig", "koi8-r", "cp866"]:
            try:
                return data.decode(e)
            except Exception:
                pass
    return data.decode("utf-8", errors="ignore")


processed = []

files = list(RAW_DIR.glob("*.txt")) + list(RAW_DIR.glob("*.TXT"))
if not files:
    print(f"Внимание: в {RAW_DIR} не найдено .txt файлов.")
    print("Положите туда текстовые версии Положений и Приказов.")

for file in sorted(files):
    text = read_any_encoding(file)
    text = text.replace("\r\n", "\n").replace("\r", "\n")

    out_txt = OUT_DIR / file.name
    out_txt.write_text(text, encoding="utf-8")

    meta = {
        "type": "internal",                   
        "name": file.stem,
        "source_file": str(file),
        "processed_file": str(out_txt),
        "text_length": len(text),
        "doc_type": "internal_regulation",
    }

    (OUT_DIR / f"{file.stem}.json").write_text(
        json.dumps(meta, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )
    processed.append(meta)

with open(OUT_DIR / "processed_index.jsonl", "w", encoding="utf-8") as f:
    for item in processed:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")

print(f"Обработано {len(processed)} внутренних документов.")
for p in processed:
    print(f"  - {p['name']} ({p['text_length']} символов)")
```
